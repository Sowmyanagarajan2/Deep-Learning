```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, SimpleRNN, Dense

# Training examples
X = np.array([
    [1, 2, 3],
    [2, 3, 4],
    [3, 4, 5],
    [4, 5, 6],
    [5, 6, 7],
    [6, 7, 8],
    [7, 8, 9],
    [8, 9, 10]
])

# Correct answers
y = np.array([4, 5, 6, 7, 8, 9, 10, 11])

# RNN needs:
# samples, time steps, features
X = X.reshape(8, 3, 1)

# Create RNN
model = Sequential([
    Input(shape=(3, 1)),
    SimpleRNN(20),
    Dense(1)
])

# Prepare model
model.compile(
    optimizer="adam",
    loss="mse"
)

# Train model
model.fit(
    X,
    y,
    epochs=500,
    verbose=0
)

# New sequence
test = np.array([5, 6, 7]).reshape(1, 3, 1)

# Predict
prediction = model.predict(test, verbose=0)

print("Predicted:", prediction[0][0])
```

```python
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Our sentences
sentences = [
    "I love this movie",
    "I hate this movie",
    "I love this film",
    "I hate this film"
]

# 1 = Positive, 0 = Negative
labels = np.array([1, 0, 1, 0])


# Convert words into numbers automatically
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)

X = tokenizer.texts_to_sequences(sentences)

# Make all sentences the same length
X = pad_sequences(X)

# Create LSTM
model = Sequential([
    Embedding(20, 8),
    LSTM(10),
    Dense(1, activation="sigmoid")
])

# Prepare model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy"
)
# Train
model.fit(X, labels, epochs=100, verbose=0)

# New sentence
test = tokenizer.texts_to_sequences(["I love this film"])
test = pad_sequences(test, maxlen=X.shape[1])


# Predict
result = model.predict(test, verbose=0)

print("Positive" if result[0][0] > 0.5 else "Negative")
```

Tokenizer converts words into numbers. Embedding represents those words. LSTM reads the sequence and remembers useful information. Dense gives the final answer. Sigmoid converts that answer into a value between 0 and 1. Loss measures the mistake, and Adam helps correct it
